```python
import os

# Works in VS Code/local Jupyter kernels (not tied to Colab UI secrets).
api_key = os.getenv("GEMINI_API_KEY")
if not api_key:
    raise RuntimeError(
        "GEMINI_API_KEY is not set. In VS Code, export it before starting the kernel."
    )

from google import genai

client = genai.Client(api_key=api_key)
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="What is the capital of France?"
 )
print(response.text)
```
